```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
matplotlib.rcParams["mathtext.fontset"] = "cm"
matplotlib.rcParams["font.family"] = "STIXGeneral"

E_tot_singlet=np.array([-1.50087067e-04, -6.17555094e-05, -9.35987445e-06, -4.80268029e-19,
       -4.80266544e-19]) # singlet

D_singlet=np.array([1.12943920e-03-2.86418093e-13j, 7.62610979e-04-1.58200149e-13j,
       3.64780182e-04+1.38896098e-12j, 3.85283078e-14-7.61713964e-18j,
       3.85283078e-14-7.61713964e-18j])
### 
E_tot_triplet=np.array([-2.04194602e-03, -1.43832364e-03, -8.62650395e-04, -3.41238395e-04,
       -2.52428925e-06])

D_triplet=np.array([0.00277063+2.00291588e-14j, 0.00217593+4.60581424e-13j,
       0.00158156+2.45291834e-13j, 0.00097203-2.76516484e-12j,
       0.00019924-5.46041059e-13j]);



Varray=np.array([-0.005, -0.004, -0.003, -0.002, -0.001]);
# (Optional) sort by V just in case
idx = np.argsort(Varray)
V = Varray[idx]
Es = E_tot_singlet[idx]
Et = E_tot_triplet[idx]
Ds = np.abs(D_singlet[idx])
Dt = np.abs(D_triplet[idx])

fig, ax = plt.subplots(figsize=(4.7*1.3, 3.8*1.3))

# Main: E vs V
ax.plot(V*1000, Es*1000, "o-", markersize=8,label="Singlet")
ax.plot(V*1000, Et*1000, "s--", markersize=8, label="Triplet")
 

ax.set_xlabel(r"$V$ (meV)", fontsize=20)
ax.set_ylabel(r"$\Omega-\Omega_N$ (meV)", fontsize=20)
ax.set_yticks([0,-1,-2]); ax.tick_params(labelsize=15)
ax.set_xticks([-5,-3, -1]);   ax.tick_params(labelsize=15)
#ax.set_title(r"$\Delta(V)$", fontsize=20)
ax.legend(fontsize=17,bbox_to_anchor=(0.0, 0.58),
                bbox_transform=ax.transAxes, loc="center left")
 
# Inset: |Delta| vs V
axins = inset_axes(
    ax,
    width=2, height=1.7,  # inches
    bbox_to_anchor=(0.56, 0.13),
    bbox_transform=ax.transAxes,
    loc="lower left",
    borderpad=0.0
)
axins.plot(V*1000, Ds*1000, "o-", label="|Δ| singlet")
axins.plot(V*1000, Dt*1000, "s--", label="|Δ| triplet")
#axins.set_title("|Δ|", fontsize=10)
#axins.set_xlabel(r"$V$ ", fontsize=17)
axins.set_ylabel(r"$|\Delta|$ (meV)", fontsize=17)
axins.set_yticks([0,1,2]); axins.tick_params(labelsize=13)
axins.set_xticks([-5,-3, -1]);   axins.tick_params(labelsize=13)
axins.grid(True, which="both", alpha=0.25)


fig.savefig("figs/fig3.pdf", bbox_inches="tight")


```
